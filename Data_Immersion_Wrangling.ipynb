```python
import pandas as pd
import numpy as np

# Correct working dataset URL
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

# Load dataset
df = pd.read_csv(url)

# Show first rows
print("First 5 Rows")
print(df.head())

# Dataset Info
print("\nDataset Info")
print(df.info())

# Missing Values
print("\nMissing Values")
print(df.isnull().sum())

# Duplicate Rows
print("\nDuplicate Rows")
print(df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

# Rename columns
df.columns = [
    "SepalLength",
    "SepalWidth",
    "PetalLength",
    "PetalWidth",
    "Species"
]

# Standardize text column
df["Species"] = df["Species"].str.strip().str.title()

# Feature Engineering
df["PetalCategory"] = pd.cut(
    df["PetalLength"],
    bins=[0,2,5,10],
    labels=["Small","Medium","Large"]
)

# Save cleaned dataset
df.to_csv("cleaned_data.csv", index=False)

print("\nCleaned Dataset")
print(df.head())

print("\nProject Completed Successfully!")
```
